```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import transform, col, concat, lit
```

```python
spark = (
    SparkSession.builder
    .appName("basic-transformations-02")
    .master("spark://spark-master:7077")
    .config("spark.executor.memory", "512m")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("ERROR")
```

```python
df = (
    spark.read.format("json")
    .option("multiLine", "true")
    .load("../data/nobel_prizes.json")
)
```

```python
df.printSchema()
```

```python
df_transformed = (
    df.select(
         "category"
        ,"overallMotivation"
        ,"year"
        ,"laureates"
        ,transform(col("laureates"), lambda x: concat(x.firstname, lit(" "), x.surname)).getItem(0)
        .alias("laureates_full_name")
    )
)
```

```python
df_transformed.show()
```

```python
df_deduped = df.dropDuplicates([
    "category",
    "overallMotivation",
    "year"
])
```

```python
df_deduped.show()
```

```python
df_sorted = df.orderBy("year")
df_sorted.show()
```

```python
df_sorted = df.orderBy(["year", "category"], ascedning=[False, True])
df_sorted.show()
```

```python
df_sorted = df.sort(["year", "category"], ascedning=[False, True])
df_sorted.show()
```

```python
df_renamed = df.withColumnRenamed("category", "Topic")
df_renamed.show()
```

```python
df_renamed = (
    df.selectExpr(
        "category as Topic"
          , "year as Year"
          , "overallMotivation as Motivation"
    )
)

df_renamed.show()
```
